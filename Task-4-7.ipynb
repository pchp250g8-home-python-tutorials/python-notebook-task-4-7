На заводе работают токари и слесари, число которых относится как (11/12) : (1/2).
Сколько всего рабочих на заводе, если токарей на 95 больше, чем слесарей?

```python
from sympy.abc import x
from sympy import Eq
from sympy import solve

ans = solve(Eq((x + 95) / x, (11 / 12) / (1 / 2)), x)
w1 = ans.pop()
w2 = w1 + 95
ans  = w1 + w2
print(w1)
print(w2)
print(ans)
```

```python
from sympy.abc import x, y
from sympy import Eq
from sympy import solve

ans = solve([Eq(x - y, 95), Eq(x / y, (11/12) / (1 / 2))], [x, y])
print(ans)
print(ans[x] + ans[y])
```
